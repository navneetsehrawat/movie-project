```python
# -------------------------------------------------
# cell 1: Import Libraries
# ---------------------------------------------------
# pandas: for data manipulation
# matplotlib & seaborn : for visualization

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# optional: set seaborn atyle for charts
sns.set(style="whitegrid")
```

```python
# -----------------------------
# cell 2 : Load dataset
# -------------------------------------
# Load the CSV file into a pandas Dataframe

df = pd.read_csv('tmdb_5000_movies.csv')

# Display first 5 rows to understand the data
df.head()
```

```python
# --------------------------
# cell 3: Data Info & Cleaning
# ------------------------------------
# Check the shape of the dataset
print("dataset Shape:", df.shape)

# check column names and types
df.info()

# check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Drop rows with missing 'budget' or 'revenue' for analysis
df_clean = df.dropna(subset=['budget','revenue'])
print("\nCleaned Dataset Shape:", df_clean.shape)
```

```python
# ------------------------------------
# cell 4: Distribution of movie budgets
# ----------------------------------------
# Visualize how movie budgets are distributed

plt.figure(figsize=(10,6))
sns.histplot(df_clean['budget'], bins=50, kde=True, color='skyblue')
plt.title('Distribution of movie budgets')
plt.xlabel('Budget')
plt.ylabel('number of Movies')
plt.show()
```

```python
# --------------------------------------------------
# cell 5 : budget vs revenue scatter plot
# --------------------------------------------------
# Explore the relationship between budget and revenue

plt.figure(figsize=(10,6))
sns.scatterplot(x='budget', y='revenue', data=df_clean, color='purple')
plt.title('Budget vs Revenue')
plt.xlabel('Budget')
plt.ylabel('revenue')
plt.show()
```

```python
# ------------------------------
# cell 6: Average rating per genre
# ---------------------------------------
# The 'genres' column may contain multiple genres seprated by
# We'll split them all and analyze ratings per genre

# Split genres into a list
df_clean['genres'] = df_clean['genres'].str.split('|')

# Explode the list into seperate rows for analysis
df_exploded = df_clean.explode('genres')

# Calculate average rating per genre
avg_rating_genre = df_exploded.groupby('genres')['vote_average'].mean().sort_values(ascending=False)

# Display the result
print("Average Rating per Genre:")
print(avg_rating_genre)

# plot bar chart
plt.figure(figsize=(12,6))
avg_rating_genre.plot(kind='bar' , color='green')
plot.title('Average Rating per Genre')
plt.xlabel('Genre')
plt.ylabel('Average Rating')
plt.xticks(rotation=45)
plt.show()
```

```python
# -----------------------------
# Cell 8: Genre Distribution Pie Chart
# -----------------------------
# Show how many movies are present per genre

genre_counts = df_exploded['genres'].value_counts()

plt.figure(figsize=(8,8))
plt.pie(genre_counts, labels=genre_counts.index, autopct='%1.1f%%', startangle=140)
plt.title('Genre Distribution of Movies')
plt.show()
```

```python
# -----------------------------
# Cell 10: Genre Distribution Donut Chart
# -----------------------------
# Donut chart is a pie chart with a hole in the center

plt.figure(figsize=(8,8))
# Pie chart
plt.pie(genre_counts, labels=genre_counts.index, autopct='%1.1f%%', startangle=140, wedgeprops={'width':0.4})
plt.title('Genre Distribution of Movies (Donut Chart)')
plt.show
```
